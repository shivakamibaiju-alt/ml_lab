# K-Means Clustering on the Digits Dataset

This notebook applies K-Means clustering to the handwritten digits dataset and compares different values of K using inertia and silhouette score.

```python
from sklearn.datasets import load_digits
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
```

## 1. Load the dataset

```python
digits = load_digits()
X = digits.data
print("Dataset shape:", X.shape)
```

## 2. Standardize the data

```python
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

## 3. Apply K-Means for different K values

```python
k_values = [8, 10, 12]

for k in k_values:
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )
    labels = kmeans.fit_predict(X_scaled)
    inertia = kmeans.inertia_
    silhouette = silhouette_score(X_scaled, labels)

    print("K =", k)
    print("Inertia =", round(inertia, 2))
    print("Silhouette Score =", round(silhouette, 4))
    print()
```

## Conclusion

- **Inertia** measures the sum of squared distances between samples and their assigned cluster centers. Lower values generally indicate more compact clusters.
- **Silhouette score** measures how well-separated the clusters are. A higher score generally indicates better-defined clusters.
- Compare both metrics for K = 8, 10, and 12 when evaluating the clustering.
